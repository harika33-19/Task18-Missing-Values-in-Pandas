```python

import pandas as pd

# 1. Create a sample dataset
data = {
    "Name": ["Ravi", "Sita", "Arjun", "Priya", "Rahul"],
    "Age": [18, None, 20, None, 19],
    "Marks": [85, 90, None, 80, 75],
    "City": ["Guntur", "Vijayawada", None, "Hyderabad", "Guntur"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 2. Identify missing values
print("\nMissing Values:")
print(df.isna())

# 3. Count missing values in each column
print("\nMissing Value Summary:")
print(df.isna().sum())

# 4. Count total missing values
print("\nTotal Missing Values:")
print(df.isna().sum().sum())

# 5. Display available ages
print("\nRows with Available Age:")
print(df[df["Age"].notna()])

# 6. Remove rows containing missing values
dropped_df = df.dropna()
print("\nDataset After Dropping Missing Rows:")
print(dropped_df)

# 7. Fill missing values
clean_df = df.copy()

clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].mean()
)
clean_df["Marks"] = clean_df["Marks"].fillna(
    clean_df["Marks"].median()
)
clean_df["City"] = clean_df["City"].fillna(
    clean_df["City"].mode()[0]
)

print("\nCleaned Dataset:")
print(clean_df)

# 8. Verify that missing values are handled
print("\nMissing Values After Cleaning:")
print(clean_df.isna().sum())

# 9. Save the cleaned dataset
clean_df.to_csv("cleaned_dataset.csv", index=False)
print("\nCleaned dataset saved successfully!")
```
